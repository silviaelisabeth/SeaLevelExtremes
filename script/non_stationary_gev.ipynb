- **Time-Varying Parameter:** Instead of fitting separate distributions per year, fit a *non-stationary* GEV over the full 1960–2019 series at each site. Typically one lets the location parameter vary linearly with time (μ(t) = μ₀ + α·t) while keeping scale and shape constant (or also varying them with time/covariates). The parameter α then measures the trend directly.
- **Fitting Method:** This can be done by maximizing the likelihood of the entire time series under the GEV with time-dependent μ. Python does not have a built-in non-stationary GEV, but one can code the likelihood and use **`scipy.optimize`** or use specialized libraries (e.g. **`pyextremes`** or even R’s **`evd`**/**`ismev`**). The estimate α (and its uncertainty) quantifies the trend in the extreme distribution’s center. Such non-stationary fits have been used to link GEV parameters to time or climate indices.
- **Interpretation:** A significant time dependence indicates changing climate conditions. For example, allowing μ to grow with time captures mean sea-level rise or increasing storm intensity. If successful, this approach often yields more robust trend estimates than fitting small samples year-by-year.
